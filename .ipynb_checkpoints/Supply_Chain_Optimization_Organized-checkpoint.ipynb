## Project Overview

This project combines machine learning and mathematical optimization to support supply chain planning. A synthetic network of suppliers, warehouses, and customers is used to explore demand patterns, allocate limited supply, and evaluate capacity scenarios.

**Workflow:** data preparation and exploration, mixed-integer linear programming, capacity sensitivity analysis, demand forecasting with Random Forest, and forecast-driven optimization.

**Data note:** all data in this project is synthetic and is intended for a portfolio proof of concept, not as a representation of a real company.

## Phase 1: Data Preparation and Exploratory Analysis

Define the supplier, warehouse, and customer network; generate the 30-day demand dataset; and examine demand distribution, customer priorities, supplier capacity, warehouse throughput, and network structure.

```python
warehouse_data = {
    "warehouse_id": ["W1", "W2", "W3", "W4", "W5"],
    "throughput_capacity_units_per_day": [800, 1000, 700, 1200, 900],
    "fixed_operating_cost_per_day": [500, 600, 450, 700, 550]
}

warehouses = pd.DataFrame(warehouse_data)

display(warehouses)

print("\nWAREHOUSE DATA SUMMARY")
print("=" * 50)

print(f"Number of warehouses: {len(warehouses)}")

print(
    "Total daily processing capacity:",
    warehouses["throughput_capacity_units_per_day"].sum()
)

print(
    "Average daily operating cost:",
    warehouses["fixed_operating_cost_per_day"].mean()
)

assert warehouses["warehouse_id"].is_unique
assert warehouses["throughput_capacity_units_per_day"].gt(0).all()
assert warehouses["fixed_operating_cost_per_day"].gt(0).all()

print("\nWarehouse data validation completed successfully.")
```

```python
days = np.arange(1, PLANNING_HORIZON + 1)

demand_records = []

for _, customer in customers.iterrows():
    base_demand = customer["average_daily_demand"]

    trend = np.linspace(0, np.random.uniform(-5, 5), PLANNING_HORIZON)

    weekly_pattern = 5 * np.sin(2 * np.pi * days / 7)

    random_noise = np.random.normal(0, 5, PLANNING_HORIZON)

    daily_demand = (
        base_demand
        + trend
        + weekly_pattern
        + random_noise
    )

    daily_demand = np.maximum(1, np.round(daily_demand)).astype(int)

    for day, demand in zip(days, daily_demand):
        demand_records.append({
            "day": day,
            "customer_id": customer["customer_id"],
            "demand": demand
        })

daily_demand = pd.DataFrame(demand_records)

display(daily_demand.head(10))

print("\nDAILY DEMAND DATA SUMMARY")
print("=" * 50)

print(f"Number of records: {len(daily_demand)}")
print(f"Number of days: {daily_demand['day'].nunique()}")
print(f"Number of customers: {daily_demand['customer_id'].nunique()}")
print(f"Total demand over 30 days: {daily_demand['demand'].sum()} units")
print(f"Average daily total demand: "
      f"{daily_demand.groupby('day')['demand'].sum().mean():.2f} units")

assert len(daily_demand) == PLANNING_HORIZON * NUM_CUSTOMERS
assert daily_demand["demand"].gt(0).all()
assert daily_demand.groupby("day").size().eq(NUM_CUSTOMERS).all()

print("\nDaily demand data validation completed successfully.")
```

## Phase 2: Baseline Supply Chain Optimization

Build a mixed-integer linear programming model with PuLP. The objective combines supplier purchasing, transportation, warehouse operating, and unmet-demand penalty costs. Constraints represent supplier capacity, warehouse flow balance and throughput, and customer demand fulfillment.

```python
model = pulp.LpProblem(
    "Supply_Chain_Optimization",
    pulp.LpMinimize
)

x = {
    (s, w): model.add_variable(
        f"SupplierToWarehouse_{s}_{w}",
        lowBound=0
    )
    for s in supplier_ids
    for w in warehouse_ids
}

y = {
    (w, c): model.add_variable(
        f"WarehouseToCustomer_{w}_{c}",
        lowBound=0
    )
    for w in warehouse_ids
    for c in customer_ids
}

u = {
    c: model.add_variable(
        f"UnmetDemand_{c}",
        lowBound=0
    )
    for c in customer_ids
}

z = {
    w: model.add_variable(
        f"WarehouseActive_{w}",
        cat="Binary"
    )
    for w in warehouse_ids
}

model += (
    pulp.lpSum(
        (float(supplier_purchase_cost.loc[s])
         + float(supplier_warehouse_cost.loc[s, w])) * x[(s, w)]
        for s in supplier_ids
        for w in warehouse_ids
    )
    + pulp.lpSum(
        float(warehouse_customer_cost.loc[w, c]) * y[(w, c)]
        for w in warehouse_ids
        for c in customer_ids
    )
    + pulp.lpSum(
        float(warehouse_fixed_cost.loc[w]) * z[w]
        for w in warehouse_ids
    )
    + pulp.lpSum(
        float(customer_shortage_penalty.loc[c]) * u[c]
        for c in customer_ids
    )
)

for s in supplier_ids:
    model += (
        pulp.lpSum(x[(s, w)] for w in warehouse_ids)
        <= float(supplier_capacity.loc[s]),
        f"SupplierCapacity_{s}"
    )

for w in warehouse_ids:
    inbound = pulp.lpSum(
        x[(s, w)] for s in supplier_ids
    )

    outbound = pulp.lpSum(
        y[(w, c)] for c in customer_ids
    )

    model += (
        inbound == outbound,
        f"WarehouseFlowBalance_{w}"
    )

    model += (
        inbound <= float(warehouse_capacity.loc[w]) * z[w],
        f"WarehouseCapacity_{w}"
    )

for c in customer_ids:
    model += (
        pulp.lpSum(y[(w, c)] for w in warehouse_ids)
        + u[c] == float(customer_total_demand.loc[c]),
        f"DemandBalance_{c}"
    )

constraint_names = [
    constraint.name for constraint in model.constraints()
]

print("Clean optimization model created.")
print("Decision variables:", len(x) + len(y) + len(u) + len(z))
print("Total constraints:", len(constraint_names))
```

```python

supplier_allocation = pd.DataFrame(
    [
        {
            "supplier_id": s,
            "warehouse_id": w,
            "quantity": x[(s, w)].value()
        }
        for s in supplier_ids
        for w in warehouse_ids
    ]
)

warehouse_allocation = pd.DataFrame(
    [
        {
            "warehouse_id": w,
            "customer_id": c,
            "quantity": y[(w, c)].value()
        }
        for w in warehouse_ids
        for c in customer_ids
    ]
)

customer_results = customers[
    ["customer_id", "priority"]
].copy()

customer_results["demand"] = customer_results[
    "customer_id"
].map(customer_total_demand)

customer_results["delivered"] = customer_results[
    "customer_id"
].map(
    lambda c: sum(
        y[(w, c)].value() or 0
        for w in warehouse_ids
    )
)

customer_results["unmet_demand"] = customer_results[
    "customer_id"
].map(
    lambda c: u[c].value() or 0
)

customer_results["fill_rate"] = (
    customer_results["delivered"]
    / customer_results["demand"]
    * 100
)

print("Supplier allocation:")
display(supplier_allocation.head())

print("\nCustomer fulfillment:")
display(customer_results.head(10))
```

```python

total_demand = customer_results["demand"].sum()
total_delivered = customer_results["delivered"].sum()
total_unmet = customer_results["unmet_demand"].sum()

overall_fill_rate = (
    total_delivered / total_demand * 100
)

supplier_usage = supplier_allocation.groupby(
    "supplier_id"
)["quantity"].sum()

supplier_utilization = (
    supplier_usage / supplier_capacity * 100
)

warehouse_usage = supplier_allocation.groupby(
    "warehouse_id"
)["quantity"].sum()

warehouse_status = pd.DataFrame({
    "warehouse_id": warehouse_ids,
    "inbound_quantity": [
        warehouse_usage.get(w, 0)
        for w in warehouse_ids
    ],
    "active": [
        int(round(z[w].value() or 0))
        for w in warehouse_ids
    ]
})

print("SUPPLY CHAIN PERFORMANCE")
print("------------------------------")
print(f"Total demand: {total_demand:,.0f}")
print(f"Total delivered: {total_delivered:,.0f}")
print(f"Total unmet demand: {total_unmet:,.0f}")
print(f"Overall fill rate: {overall_fill_rate:.2f}%")

print("\nSUPPLIER UTILIZATION")
display(supplier_utilization.rename("utilization_percent").to_frame())

print("\nWAREHOUSE STATUS")
display(warehouse_status)
```

```python
purchase_cost_total = sum(
    float(supplier_purchase_cost.loc[s]) * x[(s, w)].value()
    for s in supplier_ids
    for w in warehouse_ids
)

inbound_transport_cost = sum(
    float(supplier_warehouse_cost.loc[s, w]) * x[(s, w)].value()
    for s in supplier_ids
    for w in warehouse_ids
)

outbound_transport_cost = sum(
    float(warehouse_customer_cost.loc[w, c]) * y[(w, c)].value()
    for w in warehouse_ids
    for c in customer_ids
)

fixed_warehouse_cost = sum(
    float(warehouse_fixed_cost.loc[w]) * z[w].value()
    for w in warehouse_ids
)

shortage_penalty_cost = sum(
    float(customer_shortage_penalty.loc[c]) * u[c].value()
    for c in customer_ids
)

cost_breakdown = pd.DataFrame({
    "cost_component": [
        "Supplier Purchase",
        "Inbound Transportation",
        "Outbound Transportation",
        "Warehouse Fixed Cost",
        "Shortage Penalty"
    ],
    "total_cost": [
        purchase_cost_total,
        inbound_transport_cost,
        outbound_transport_cost,
        fixed_warehouse_cost,
        shortage_penalty_cost
    ]
})

cost_breakdown["percentage"] = (
    cost_breakdown["total_cost"]
    / cost_breakdown["total_cost"].sum()
    * 100
)

display(cost_breakdown.round(2))

print("Calculated Total Cost:",
      round(cost_breakdown["total_cost"].sum(), 2))

print("Solver Objective Value:",
      round(solve_stats.objective, 2))
```

## Phase 3: Historical-Demand Capacity Sensitivity Analysis

Evaluate how increasing supplier capacity changes delivered quantity, unmet demand, fill rate, and total modeled cost. Capacity expansion investment costs are not included, so lower modeled operating cost should not be interpreted as proof of investment profitability.

```python

def solve_capacity_scenario(capacity_multiplier):

    scenario_model = pulp.LpProblem(
        "Supply_Chain_Scenario",
        pulp.LpMinimize
    )

    scenario_x = {
        (s, w): scenario_model.add_variable(
            f"x_{s}_{w}", lowBound=0
        )
        for s in supplier_ids
        for w in warehouse_ids
    }

    scenario_y = {
        (w, c): scenario_model.add_variable(
            f"y_{w}_{c}", lowBound=0
        )
        for w in warehouse_ids
        for c in customer_ids
    }

    scenario_u = {
        c: scenario_model.add_variable(
            f"u_{c}", lowBound=0
        )
        for c in customer_ids
    }

    scenario_z = {
        w: scenario_model.add_variable(
            f"z_{w}", cat="Binary"
        )
        for w in warehouse_ids
    }

    scenario_model += (
        pulp.lpSum(
            (
                float(supplier_purchase_cost.loc[s])
                + float(supplier_warehouse_cost.loc[s, w])
            ) * scenario_x[(s, w)]
            for s in supplier_ids
            for w in warehouse_ids
        )
        + pulp.lpSum(
            float(warehouse_customer_cost.loc[w, c])
            * scenario_y[(w, c)]
            for w in warehouse_ids
            for c in customer_ids
        )
        + pulp.lpSum(
            float(warehouse_fixed_cost.loc[w]) * scenario_z[w]
            for w in warehouse_ids
        )
        + pulp.lpSum(
            float(customer_shortage_penalty.loc[c]) * scenario_u[c]
            for c in customer_ids
        )
    )

    for s in supplier_ids:
        scenario_model += (
            pulp.lpSum(
                scenario_x[(s, w)] for w in warehouse_ids
            )
            <= float(supplier_capacity.loc[s]) * capacity_multiplier
        )

    for w in warehouse_ids:
        inbound = pulp.lpSum(
            scenario_x[(s, w)] for s in supplier_ids
        )

        outbound = pulp.lpSum(
            scenario_y[(w, c)] for c in customer_ids
        )

        scenario_model += inbound == outbound

        scenario_model += (
            inbound <= float(warehouse_capacity.loc[w])
            * scenario_z[w]
        )

    for c in customer_ids:
        scenario_model += (
            pulp.lpSum(
                scenario_y[(w, c)] for w in warehouse_ids
            )
            + scenario_u[c]
            == float(customer_total_demand.loc[c])
        )

    stats = scenario_model.solve(
        pulp.COIN_CMD(msg=False)
    )

    total_delivered = sum(
        scenario_y[(w, c)].value() or 0
        for w in warehouse_ids
        for c in customer_ids
    )

    total_unmet = sum(
        scenario_u[c].value() or 0
        for c in customer_ids
    )

    total_demand = float(total_30_day_demand)

    return {
        "capacity_increase_percent": (
            capacity_multiplier - 1
        ) * 100,
        "total_capacity": (
            float(supplier_capacity.sum())
            * capacity_multiplier
        ),
        "total_delivered": total_delivered,
        "total_unmet": total_unmet,
        "fill_rate_percent": (
            total_delivered / total_demand
        ) * 100,
        "total_cost": stats.objective,
        "solver_status": stats.status_str
    }


scenario_multipliers = [1.0, 1.25, 1.5, 2.0]

scenario_results = pd.DataFrame(
    [
        solve_capacity_scenario(multiplier)
        for multiplier in scenario_multipliers
    ]
)

display(scenario_results.round(2))
```

```python

def scenario_cost_breakdown(capacity_multiplier):

    scenario_model = pulp.LpProblem(
        "Scenario_Cost_Analysis",
        pulp.LpMinimize
    )

    sx = {
        (s, w): scenario_model.add_variable(
            f"sx_{s}_{w}", lowBound=0
        )
        for s in supplier_ids for w in warehouse_ids
    }

    sy = {
        (w, c): scenario_model.add_variable(
            f"sy_{w}_{c}", lowBound=0
        )
        for w in warehouse_ids for c in customer_ids
    }

    su = {
        c: scenario_model.add_variable(
            f"su_{c}", lowBound=0
        )
        for c in customer_ids
    }

    sz = {
        w: scenario_model.add_variable(
            f"sz_{w}", cat="Binary"
        )
        for w in warehouse_ids
    }

    scenario_model += (
        pulp.lpSum(
            (float(supplier_purchase_cost.loc[s])
             + float(supplier_warehouse_cost.loc[s, w]))
            * sx[(s, w)]
            for s in supplier_ids for w in warehouse_ids
        )
        + pulp.lpSum(
            float(warehouse_customer_cost.loc[w, c])
            * sy[(w, c)]
            for w in warehouse_ids for c in customer_ids
        )
        + pulp.lpSum(
            float(warehouse_fixed_cost.loc[w]) * sz[w]
            for w in warehouse_ids
        )
        + pulp.lpSum(
            float(customer_shortage_penalty.loc[c]) * su[c]
            for c in customer_ids
        )
    )

    for s in supplier_ids:
        scenario_model += (
            pulp.lpSum(sx[(s, w)] for w in warehouse_ids)
            <= float(supplier_capacity.loc[s]) * capacity_multiplier
        )

    for w in warehouse_ids:
        inbound = pulp.lpSum(
            sx[(s, w)] for s in supplier_ids
        )
        outbound = pulp.lpSum(
            sy[(w, c)] for c in customer_ids
        )

        scenario_model += inbound == outbound
        scenario_model += (
            inbound <= float(warehouse_capacity.loc[w]) * sz[w]
        )

    for c in customer_ids:
        scenario_model += (
            pulp.lpSum(
                sy[(w, c)] for w in warehouse_ids
            ) + su[c]
            == float(customer_total_demand.loc[c])
        )

    stats = scenario_model.solve(
        pulp.COIN_CMD(msg=False)
    )

    values = {
        "Supplier Purchase": sum(
            float(supplier_purchase_cost.loc[s]) * sx[(s, w)].value()
            for s in supplier_ids for w in warehouse_ids
        ),
        "Inbound Transportation": sum(
            float(supplier_warehouse_cost.loc[s, w]) * sx[(s, w)].value()
            for s in supplier_ids for w in warehouse_ids
        ),
        "Outbound Transportation": sum(
            float(warehouse_customer_cost.loc[w, c]) * sy[(w, c)].value()
            for w in warehouse_ids for c in customer_ids
        ),
        "Warehouse Fixed Cost": sum(
            float(warehouse_fixed_cost.loc[w]) * sz[w].value()
            for w in warehouse_ids
        ),
        "Shortage Penalty": sum(
            float(customer_shortage_penalty.loc[c]) * su[c].value()
            for c in customer_ids
        )
    }

    return {
        "capacity_increase_percent": (capacity_multiplier - 1) * 100,
        "solver_status": stats.status_str,
        **values,
        "total_cost": sum(values.values())
    }


cost_comparison = pd.DataFrame([
    scenario_cost_breakdown(1.4),
    scenario_cost_breakdown(1.5)
])

display(cost_comparison.round(2))
```

## Phase 4: Demand Forecasting with Random Forest

Create lag-based and calendar features, use a chronological train-test split, and compare a Random Forest regressor with a seven-day lag baseline. Generate a 30-day customer-level demand forecast for the optimization stage.

```python

forecast_data = daily_demand.sort_values(
    ["customer_id", "day"]
).copy()

forecast_data["lag_1"] = (
    forecast_data.groupby("customer_id")["demand"].shift(1)
)

forecast_data["lag_7"] = (
    forecast_data.groupby("customer_id")["demand"].shift(7)
)

forecast_data["rolling_mean_7"] = (
    forecast_data.groupby("customer_id")["demand"]
    .transform(lambda s: s.shift(1).rolling(7).mean())
)

forecast_data["day_of_week"] = (
    (forecast_data["day"] - 1) % 7
)

forecast_data = forecast_data.merge(
    customers[["customer_id", "average_daily_demand"]],
    on="customer_id",
    how="left"
)

forecast_data = forecast_data.dropna().reset_index(drop=True)

print("Forecast dataset shape:", forecast_data.shape)

display(forecast_data.head(10))

print("\nDate range:")
print(
    "From day",
    forecast_data["day"].min(),
    "to day",
    forecast_data["day"].max()
)

print("\nMissing values:")
display(forecast_data.isnull().sum())
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd


train_data = forecast_data[forecast_data["day"] <= 23].copy()
test_data = forecast_data[forecast_data["day"] >= 24].copy()

features = [
    "lag_1",
    "lag_7",
    "rolling_mean_7",
    "day_of_week",
    "average_daily_demand"
]

X_train = train_data[features]
y_train = train_data["demand"]

X_test = test_data[features]
y_test = test_data["demand"]

# Train Random Forest model

rf_model = RandomForestRegressor(
    n_estimators=200,
    max_depth=8,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)


rf_predictions = rf_model.predict(X_test)


baseline_predictions = test_data["lag_7"].values

def evaluate_model(actual, predicted, model_name):
    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    wape = (
        np.sum(np.abs(actual - predicted))
        / np.sum(actual)
    ) * 100

    return {
        "model": model_name,
        "MAE": round(mae, 2),
        "RMSE": round(rmse, 2),
        "WAPE (%)": round(wape, 2)
    }

evaluation_results = pd.DataFrame([
    evaluate_model(y_test, baseline_predictions, "7-Day Lag Baseline"),
    evaluate_model(y_test, rf_predictions, "Random Forest")
])

display(evaluation_results)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python

prediction_results = test_data[
    ["day", "customer_id", "demand"]
].copy()

prediction_results["rf_prediction"] = np.maximum(
    rf_predictions, 0
)

daily_prediction_results = (
    prediction_results
    .groupby("day")
    .agg(
        actual_demand=("demand", "sum"),
        predicted_demand=("rf_prediction", "sum")
    )
    .reset_index()
)

plt.figure(figsize=(13, 6))

plt.plot(
    daily_prediction_results["day"],
    daily_prediction_results["actual_demand"],
    marker="o",
    label="Actual Demand"
)

plt.plot(
    daily_prediction_results["day"],
    daily_prediction_results["predicted_demand"],
    marker="s",
    label="Random Forest Prediction"
)

plt.title("Actual vs Predicted Total Daily Demand")
plt.xlabel("Day")
plt.ylabel("Total Demand")
plt.xticks(range(24, 31))
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    "figures/demand_forecast_actual_vs_predicted.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

display(daily_prediction_results)
```

## Phase 5: Forecast-Driven Supply Chain Optimization

Feed the customer-level demand forecast into a second optimization model while retaining the same network structure, cost components, and operational constraints. Compare the resulting allocation and fulfillment with the historical-demand model.

```python

supplier_capacity = supplier_capacity.astype(float)
warehouse_capacity = warehouse_capacity.astype(float)

warehouse_fixed_cost = warehouse_fixed_cost.astype(float)
supplier_purchase_cost = supplier_purchase_cost.astype(float)

customer_shortage_penalty = customer_shortage_penalty.astype(float)

print("Optimization input types corrected.")
print("Supplier capacity type:", type(supplier_capacity["S1"]))
print("Warehouse capacity type:", type(warehouse_capacity["W1"]))
```

```python

forecast_model = pulp.LpProblem(
    "Forecast_Driven_Supply_Chain_Optimization",
    pulp.LpMinimize
)


x_forecast = {
    (s, w): forecast_model.add_variable(
        f"SupplierToWarehouse_{s}_{w}",
        lowBound=0
    )
    for s in suppliers["supplier_id"]
    for w in warehouses["warehouse_id"]
}

y_forecast = {
    (w, c): forecast_model.add_variable(
        f"WarehouseToCustomer_{w}_{c}",
        lowBound=0
    )
    for w in warehouses["warehouse_id"]
    for c in customers["customer_id"]
}

u_forecast = {
    c: forecast_model.add_variable(
        f"UnmetDemand_{c}",
        lowBound=0
    )
    for c in customers["customer_id"]
}

z_forecast = {
    w: forecast_model.add_variable(
        f"WarehouseActive_{w}",
        cat="Binary"
    )
    for w in warehouses["warehouse_id"]
}


forecast_model += (
    pulp.lpSum(
        x_forecast[s, w] * (
            supplier_purchase_cost[s]
            + supplier_warehouse_cost.loc[s, w]
        )
        for s, w in x_forecast
    )
    +
    pulp.lpSum(
        y_forecast[w, c] * warehouse_customer_cost.loc[w, c]
        for w, c in y_forecast
    )
    +
    pulp.lpSum(
        z_forecast[w] * warehouse_fixed_cost[w]
        for w in z_forecast
    )
    +
    pulp.lpSum(
        u_forecast[c] * customer_shortage_penalty[c]
        for c in u_forecast
    )
)


for s in suppliers["supplier_id"]:
    forecast_model += (
        pulp.lpSum(
            x_forecast[s, w]
            for w in warehouses["warehouse_id"]
        )
        <= supplier_capacity[s]
    )



for w in warehouses["warehouse_id"]:

    forecast_model += (
        pulp.lpSum(
            x_forecast[s, w]
            for s in suppliers["supplier_id"]
        )
        ==
        pulp.lpSum(
            y_forecast[w, c]
            for c in customers["customer_id"]
        )
    )

    forecast_model += (
        pulp.lpSum(
            y_forecast[w, c]
            for c in customers["customer_id"]
        )
        <= warehouse_capacity[w] * z_forecast[w]
    )



for c in customers["customer_id"]:
    forecast_model += (
        pulp.lpSum(
            y_forecast[w, c]
            for w in warehouses["warehouse_id"]
        )
        + u_forecast[c]
        == forecast_demand[c]
    )

print("Forecast-driven model created successfully.")
print("Variables:", len(forecast_model.variables()))
print("Constraints:", len(forecast_model.constraints()))
```

```python

forecast_solver = pulp.COIN_CMD(msg=True)

forecast_solve_stats = forecast_model.solve(forecast_solver)

print("Solver status:", forecast_solve_stats.status_str)
print("Has feasible solution:", forecast_solve_stats.has_solution)

if forecast_solve_stats.has_solution:

    print(
        "Objective value:",
        round(forecast_solve_stats.objective, 2)
    )

    total_forecast_delivered = sum(
        var.value() or 0
        for var in y_forecast.values()
    )

    total_forecast_unmet = sum(
        var.value() or 0
        for var in u_forecast.values()
    )

    forecast_fill_rate = (
        total_forecast_delivered
        / forecast_demand.sum()
    ) * 100

    print("\nTotal forecast demand:", round(forecast_demand.sum(), 2))
    print("Total delivered:", round(total_forecast_delivered, 2))
    print("Total unmet demand:", round(total_forecast_unmet, 2))
    print("Fill rate (%):", round(forecast_fill_rate, 2))

else:
    print("No feasible solution was found.")
```

## Phase 6: Forecast-Based Capacity Scenarios

Repeat the capacity sensitivity analysis using forecast demand. Compare historical and forecast-driven scenarios across delivery, shortages, fill rate, and modeled cost. Differences between the two models are not, by themselves, evidence that forecasting improves financial performance.

```python

def solve_forecast_scenario(capacity_multiplier):

    model = pulp.LpProblem(
        "Forecast_Capacity_Scenario",
        pulp.LpMinimize
    )

    x = {
        (s, w): model.add_variable(
            f"x_{s}_{w}",
            lowBound=0
        )
        for s in suppliers["supplier_id"]
        for w in warehouses["warehouse_id"]
    }

    y = {
        (w, c): model.add_variable(
            f"y_{w}_{c}",
            lowBound=0
        )
        for w in warehouses["warehouse_id"]
        for c in customers["customer_id"]
    }

    u = {
        c: model.add_variable(
            f"u_{c}",
            lowBound=0
        )
        for c in customers["customer_id"]
    }

    z = {
        w: model.add_variable(
            f"z_{w}",
            cat="Binary"
        )
        for w in warehouses["warehouse_id"]
    }

    model += (
        pulp.lpSum(
            x[s, w] * (
                supplier_purchase_cost[s]
                + supplier_warehouse_cost.loc[s, w]
            )
            for s, w in x
        )
        +
        pulp.lpSum(
            y[w, c] * warehouse_customer_cost.loc[w, c]
            for w, c in y
        )
        +
        pulp.lpSum(
            z[w] * warehouse_fixed_cost[w]
            for w in z
        )
        +
        pulp.lpSum(
            u[c] * customer_shortage_penalty[c]
            for c in u
        )
    )

    for s in suppliers["supplier_id"]:
        model += (
            pulp.lpSum(
                x[s, w]
                for w in warehouses["warehouse_id"]
            )
            <= float(supplier_capacity[s]) * capacity_multiplier
        )

    for w in warehouses["warehouse_id"]:

        model += (
            pulp.lpSum(
                x[s, w]
                for s in suppliers["supplier_id"]
            )
            ==
            pulp.lpSum(
                y[w, c]
                for c in customers["customer_id"]
            )
        )

        model += (
            pulp.lpSum(
                y[w, c]
                for c in customers["customer_id"]
            )
            <= float(warehouse_capacity[w]) * z[w]
        )

    for c in customers["customer_id"]:
        model += (
            pulp.lpSum(
                y[w, c]
                for w in warehouses["warehouse_id"]
            )
            + u[c]
            == float(forecast_demand[c])
        )

    stats = model.solve(pulp.COIN_CMD(msg=False))

    if not stats.has_solution:
        return None

    delivered = sum(
        var.value() or 0
        for var in y.values()
    )

    unmet = sum(
        var.value() or 0
        for var in u.values()
    )

    return {
        "capacity_increase_percent": (capacity_multiplier - 1) * 100,
        "total_capacity": supplier_capacity.sum() * capacity_multiplier,
        "delivered": delivered,
        "unmet_demand": unmet,
        "fill_rate_percent": delivered / forecast_demand.sum() * 100,
        "total_cost": stats.objective,
        "solver_status": stats.status_str
    }
```

```python

forecast_scenarios = []

for multiplier in [1.0, 1.25, 1.5, 2.0]:
    result = solve_forecast_scenario(multiplier)

    if result is not None:
        forecast_scenarios.append(result)

forecast_scenario_results = pd.DataFrame(forecast_scenarios)

forecast_scenario_results = forecast_scenario_results.round(2)

display(forecast_scenario_results)
```

## Project Conclusion and Limitations

The project demonstrates an end-to-end workflow that links demand forecasting with supply chain allocation optimization. In the test period, Random Forest achieved lower MAE, RMSE, and WAPE than the seven-day lag baseline. The forecast was then used as demand input to a PuLP optimization model.

The analysis indicates that supplier capacity is a major bottleneck in the synthetic network. Capacity scenarios improve fulfillment, although additional capacity is not always fully utilized. Historical- and forecast-driven optimization results are broadly similar; small cost differences should not be attributed to machine learning alone because the demand inputs differ.

**Limitations:** the dataset is synthetic and covers only 30 days; the test period is short; the optimization model aggregates the planning horizon and does not simulate daily inventory dynamics or lead times; and capacity expansion costs are excluded.

**Possible future extensions:** use longer real-world demand histories, model inventory and replenishment timing, incorporate lead-time and demand uncertainty, and include capital expenditure in capacity planning.
